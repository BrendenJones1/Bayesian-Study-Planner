# Network queries
Builds the network from `cpts.json` and answers P(grade >= t | e, a) and E[stress next | e, a]. All logic is in `bsp_eval.network`.

```python
from pathlib import Path

import pandas as pd
from bsp_eval.network import answer, load_cpts

cpts = load_cpts(Path("../../priors/tables/cpts.json"))
evidence = {"M1": "High", "M2": "Med", "M3": "Med", "D1": "Hard", "D2": "Med", "D3": "Med",
            "Timeliness": "On time", "DaysLeft": "3-7", "GradedSoFar": 25, "ItemWeight": ">15%",
            "LoadToday": "Med", "LoadTomorrow": "High"}
rows = {a: answer(cpts, "MATH 100", evidence, a) for a in ["Intense", "Light", "Rest"]}
pd.DataFrame({a: r["p_grade"] for a, r in rows.items()}).rename_axis("target")
```

```python
pd.Series({a: r["e_stress_next"] for a, r in rows.items()}, name="E[stress next]")
```
