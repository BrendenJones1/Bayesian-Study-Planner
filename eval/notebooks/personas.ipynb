# Personas
Four simulated students (struggling, advanced, overloaded, far from the exam). The chosen plan is tabulated as concept difficulty varies over Easy, Med, Hard, lambda over its presets and theta over {0, 0.1, 0.2, 0.3}. Plans list one action per course.


```python
from pathlib import Path

import pandas as pd
from bsp_eval.network import load_cpts
from bsp_eval.personas import difficulty_changes_action, plan_table

rows = plan_table(load_cpts(Path('../../priors/tables/cpts.json')))
df = pd.DataFrame(rows)
df['plan'] = df['plan'].map(' + '.join)
df.pivot_table(index=['persona', 'lambda', 'theta'], columns='difficulty', values='plan', aggfunc='first')
```

```python
print('Different concept difficulties led to different actions:', difficulty_changes_action(rows))
```

**Result:** yes, but only in one persona. For the overloaded student (three exams, high load) Hard concepts move the Intense hours from the first course to the third, the one where mastery is Low; Easy and Med give the same plan. For the other three personas the plan does not depend on concept difficulty; it depends on lambda (the stress cost) instead.
